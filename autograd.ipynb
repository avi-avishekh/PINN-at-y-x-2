```python
import torch
```

```python
x=torch.linspace(0,1,100, requires_grad=True)

print("Collocation Points:")
print(x)

print("\nNumber of Points:")
print(x.shape)
```

```python
import torch.nn as nn
class PINN(nn.Module):
    
    def __init__(self):
        super().__init__()
        
        self.network=nn.Sequential(
            nn.Linear(1,50),
            nn.Tanh(),
            nn.Linear(50,50),
            nn.Tanh(),
            nn.Linear(50,1)
        )
    
    def  forward(self,x):
        return self.network(x)


##Create collocation points
x=torch.linspace(0,1,11)
x=x.reshape(-1,1)

model=PINN()

u=model(x)

print("Input shape:", x.shape)
print("Output shape:", u.shape)

print("\nNetwork Output:")
print(u)
```
